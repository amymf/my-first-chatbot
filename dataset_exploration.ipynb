```python
from datasets import load_dataset

dataset = load_dataset("daily_dialog", trust_remote_code=True)
```

```python
print(dataset["train"][0])
```

```python
train_data = dataset['train']

input_response_pairs = []

for dialog in train_data['dialog']:
    for i in range(len(dialog) - 1):
        input_utterance = dialog[i].strip()
        response_utterance = dialog[i + 1].strip()
        input_response_pairs.append((input_utterance, response_utterance))

# Print the first 5 input-response pairs
for i in range(5):
    print(f"Input: {input_response_pairs[i][0]}")
    print(f"Response: {input_response_pairs[i][1]}")
    print()
```
